```python
from optlang import Model
from json import load, dump
with open("correct_MSDB.json") as jsonIn:
    model = load(jsonIn)
    
optlang_model = Model.from_json(model)
proposed_changes = {}
for rxn in optlang_model.reactions:
    for met in rxn.metabolites:
        for ele in met.elements:
            varName = f"{rxn.id}_{met.id}~{ele}"
            original_stoich = met.elements[ele]
            proposed_stoich = optlang_model.primal_values[varName]
            if original_stoich == proposed_stoich:  continue
            proposed_changes[varName] = {"Original": original_stoich, "Proposed": proposed_stoich}
print(proposed_changes)
with open("proposed_changes.json", "w") as out:
    dump(proposed_changes, out, indent=3)
# print([index for index, val in enumerate(primals.values()) if val != 0])
```

# Heuristic to categorize compounds sugars

```python
from modelseedpy.biochem import from_local
msdb = from_local(".")
```

### sugars

```python
# https://www.ebi.ac.uk/chebi/searchId.do?chebiId=CHEBI%3A35381
# https://www.ebi.ac.uk/chebi/searchId.do?chebiId=CHEBI%3A36233
from numpy import array, save, load
sugars = array([(met.id, met.name) for met in msdb.compounds if met.name[-3:] == "ose" and "C" in met.elements and met.elements["C"] < 15])
save("categorized_sugars.npy", sugars)
# new_sugars = load("categorized_sugars.npy")
print(sugars)
# for met in msdb.compounds:
#     print(met.aliases)
```

### amino acids

```python
# https://www.ebi.ac.uk/chebi/searchId.do?chebiId=CHEBI%3A33709
aminoacids = array([("cpd00035", "Alanine"), ("cpd00051", "L-Arginine"), ("cpd00132", "L-Asparagine"), ("cpd00041", "L-Aspartate"), ("cpd00084", "L-Cysteine"), ("cpd00023", "L-Glutamate"), ("cpd00053", "L-Glutamine"), ("cpd00033", "Glycine"),
                    ("cpd00119", "L-Histidine"), ("cpd00747", "L-Hydroxyproline"), ("cpd00322", "L-Isoleucine"), ("cpd00107", "L-Leucine"), ("cpd00039", "L-Lysine"), ("cpd00060", "L-Methionine"), ("cpd00066", "L-Phenylalanine"), ("cpd00129", "L-Proline"),
                    ("cpd00054", "L-Serine"), ("cpd00161", "L-Threonine"), ("cpd00065", "L-Tryptophan"), ("cpd00069", "L-Tyrosine"), ("cpd00156", "L-Valine")])
save("aminoAcids.npy", aminoacids)
print(aminoacids)
```

### minerals

```python
import re
minerals = array([(met.id, met.name) for met in msdb.compounds if met.elements != {} and "C" not in met.elements and not re.search("^R([A-Z]|\-)", met.name)])
print(minerals)
save("minerals.npy", minerals)
```

### vitamins

```python
# https://www.ebi.ac.uk/chebi/searchId.do?chebiId=CHEBI:33229
# the multitude of variations -- such as Vitamin D2 and Vitamin K2 -- and taxa-specific auxotrophies may need to be captured by expanding the given list 
vitamins = array([("Retinol", "cpd00365"), ("L-Ascorbate", "cpd00059"), ("Pyridoxyl", "cpd00215"), ("Thiamin", "cpd00305"), ("Niacin", "cpd00218"), ("Riboflavin", "cpd00220"), ("PAN", "cpd00644"), ("Cbl", "cpd00635"), ("BIOT", "cpd00104"),
                  ("Choline", "cpd00098"), ("Vitamin D3", "cpd03226"), ("Vitamin E", "cpd01628"), ("Vitamin K1", "cpd01401"), ("Folate", "cpd00393")])
vitamins = array(list(zip(vitamins[:,1], vitamins[:,0])))
print(vitamins)
save("vitamins.npy", vitamins)
```

### energy

```python
from numpy import concatenate

energy_compounds = [("cpd11640", "H2")]
sugars_aminoAcids_vitamins = concatenate((sugars[:,0], aminoacids[:,0], vitamins[:,0]))
for met in msdb.compounds:
    if met.elements != {} and "C" in met.elements and met.id not in sugars_aminoAcids_vitamins:
        energy_compounds.append((met.id, met.name))
energy_compounds = array(energy_compounds)
print(energy_compounds)
save("energy_compounds", energy_compounds)
```
